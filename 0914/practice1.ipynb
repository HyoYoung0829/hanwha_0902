```python
from langchain_openai import ChatOpenAI
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
llm = ChatOpenAI(
    temperature=0.1,
    model="gpt-4o-mini",
    
)
```

```python
question = "미국의 수도는 어디인가요?"
```

```python
response = llm.invoke(question)
print(response)
```

```python
# logprob 로그 프로버블리티 활성화
# 모델의 답변 확신도 (컨피던스 레벨) : 0에 가까울수록 정답에 가까운 확신을 가진다.

llm_with_logprob = ChatOpenAI(
    temperature=0.1,
    max_completion_tokens=2048,
    model="gpt-4o-mini",
).bind(logprobs=True)
```

```python
question2 = "인도의 수도는 어디인가요?"

response = llm_with_logprob.invoke(question2)

response
```
